1.	Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.


```python
numbers = [1,2,3,4,5]

numbers.append(6)
print("Append:", numbers)

numbers.extend([7,8,9])
print("Extend:", numbers)

numbers.insert(9,10)
print("Insert:", numbers)

numbers.remove(1)
print("Remove:", numbers)

numbers.pop(0)
print("Pop:", numbers)

print("Index of 4:", numbers.index(4))

print("Count of 3:", numbers.count(3))

numbers.sort()
print("Sort:", numbers)

numbers.reverse()
print("Reverse:", numbers)

numbers.clear()
print("Clear:", numbers)
```

7.	Given a number, check if it's positive, and if it is also even. If not positive, print if it's zero or negative.

```python
number = int(input("Enter a number: "))
if number > 0:
    if number % 2 == 0:
        print(number, "is positive and even")
    else:
        print(number, "is positive but odd")
elif number == 0:
    print(number, "is zero")
else:
    print(number, "is negative")
```
