```python
import pandas as pd

new_patients = pd.DataFrame({
    'PatientID':[106,107],
    'Name':['Frank','Grace'],
    'Department':['Neurology','Cardiology'],
    'Doctor':['Dr. John','Dr. Smith'],
    'BillAmount':[4500,6800],
    'CheckInTime':['2023-01-15 10:00','2023-01-16 11:30'],
    'InsuranceCovered':[1500,2500],
    'FinalAmount':[3000,4300]
})

final_dataset = pd.concat([merged_dataset, new_patients], ignore_index=True)

print(final_dataset)
```

```python
new_billing_columns = pd.DataFrame({
    'InsuranceType':['Gold','Silver','Gold','Silver','Gold'],
    'PaymentStatus':['Paid','Pending','Paid','Paid','Pending']
})

final_dataset = pd.concat(
    [merged_dataset, new_billing_columns],
    axis=1
)

print(final_dataset)
```
