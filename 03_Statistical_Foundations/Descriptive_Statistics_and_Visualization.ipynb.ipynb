to remove the excess code and more manual typing

i crated a list with the feature names and make them iterate in the loop by using list comprehension

and i also used f strings in the statement

```python
num_cols = [
    'Volume',
    'Avg Price',
    'Total Sales Value',
    'Discount Rate (%)',
    'Discount Amount'
]

for col in num_cols:
    plt.figure(figsize=(8,4))

    sns.boxplot(
        data=df,
        x=col
    )

    plt.title(f'Boxplot of {col}')

    plt.show()
```

bar plots fo the categorical colummns

xticks is used for the prevention of the overlapping of the labels in the x-axsis

```python
sns.countplot(
    data= df,
    x = 'Day'
)
plt.title('distrubution in days')
plt.xlabel('days')
plt.ylabel('count')

plt.xticks(rotation=45)
plt.grid(axis ='y')
plt.show()
```

```python
categorical_cols = [
    'City',
    'BU',
    'Brand',
    'Model',
    'SKU'
]
for col in categorical_cols:
  plt.figure(figsize=(8,5))

  sns.countplot(
      data = df,
      x= col
  )
  plt.title(f"distrubution of {col}")
  plt.xlabel(col)
  plt.ylabel('count')

  plt.xticks(rotation = 45)
  plt.grid(axis = 'y')
  plt.show()
```

Descriptive statistics, histograms, boxplots and bar charts were used to analyze the dataset. Sales variables were positively skewed and had a number of high-value outliers. Net Sales Value was significantly negatively correlated with Discount Rate (%). Brand, City, BU, and Day were useful as categorical variables and Model and SKU were uniformly distributed. In summary, the analysis assisted in identifying patterns of data distribution, outliers, and important relationships in the dataset.
